## Gráfico de las señales

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Leer el archivo JSON directamente a un DataFrame de Pandas
df = pd.read_json('./build/datos.json')
df.set_index('ix', inplace=True)

# 2. Graficar todo junto
plt.figure(figsize=(64, 7))

# Graficamos la señal original más tenue
plt.plot(df['y'], label='Original', alpha=0.3, color='gray', linewidth=1.0)

# Graficamos un par de medias móviles para comparar (o puedes hacer un ciclo para las 8)
plt.plot(df['media_movil_p2'], label='Media Móvil P2', linewidth=1.2)
plt.plot(df['media_movil_p4'], label='Media Móvil P4', linewidth=1.4)
plt.plot(df['media_movil_p8'], label='Media Móvil P8', linewidth=1.6)
plt.plot(df['media_movil_p16'], label='Media Móvil P16', linewidth=1.8)
plt.plot(df['media_movil_p32'], label='Media Móvil P32', linewidth=2.0)
plt.plot(df['media_movil_p64'], label='Media Móvil P64', linewidth=2.2)
plt.plot(df['media_movil_p128'], label='Media Móvil P128', linewidth=2.4)
plt.plot(df['media_movil_p256'], label='Media Móvil P128', linewidth=2.6)

plt.title('Comparativa de Medias Móviles Concurrentes (CUDA)')
plt.xlabel('Muestras')
plt.ylabel('Amplitud')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```

## Histograma de las señales

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Leer el archivo JSON directamente a un DataFrame de Pandas
df = pd.read_json('./build/datos.json')
df.set_index('ix', inplace=True)

# 2. Filtrar los datos a partir del índice 256 para el análisis estadístico
df_estables = df.loc[256:]

# 3. Graficar todo junto
plt.figure(figsize=(12, 7))

# ====================================
# Histograma (A partir del índice 256)
# ====================================
# Calculamos dinámicamente los límites basados en la señal original (max - min) en el tramo estable
min_val = df_estables['y'].min()
max_val = df_estables['y'].max()
clases = np.linspace(min_val, max_val, 129) # 129 puntos generan exactamente 128 intervalos (clases)

# Graficamos el histograma de la señal original
plt.hist(df_estables['y'], bins=clases, alpha=0.5, color='gray', edgecolor='black', label='Original (>=256)')

# Graficamos los histogramas de las medias móviles
# Observar cómo la media móvil de 256 períodos reduce la varianza
#plt.hist(df_estables['media_movil_p2'],    bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P2 (>=256)')
#plt.hist(df_estables['media_movil_p4'],    bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P4 (>=256)')
#plt.hist(df_estables['media_movil_p8'],    bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P8 (>=256)')
plt.hist(df_estables['media_movil_p16'],   bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P16 (>=256)')
#plt.hist(df_estables['media_movil_p32'],   bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P32 (>=256)')
#plt.hist(df_estables['media_movil_p64'],   bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P64 (>=256)')
#plt.hist(df_estables['media_movil_p128'],  bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P128 (>=256)')
#plt.hist(df_estables['media_movil_p256'],  bins=clases, alpha=0.5, histtype='bar', linewidth=2, label='P256 (>=256)')

plt.title('Distribución de la Señal (128 Clases)')
plt.xlabel('Amplitud')
plt.xticks(np.arange(0, 1.1, 0.1))
plt.ylabel('Frecuencia (Conteos)')
plt.legend(loc='upper right')
plt.grid(True)

plt.tight_layout()
plt.show()
```
